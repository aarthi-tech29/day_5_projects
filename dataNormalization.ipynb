# Project 1: Data Normalization

```python
import numpy as np

# Create a 2D array (5 samples, 4 features)
arr = np.random.randint(10, 100, (5, 4))
print("Original Array:\n", arr)

# Calculate mean and standard deviation for each column
mean = arr.mean(axis=0)
std = arr.std(axis=0)

# Normalize: (value - mean) / std
normalized_arr = (arr - mean) / std

print("\nMean of each column:", mean)
print("Standard Deviation of each column:", std)
print("\nNormalized Array:\n", normalized_arr)
```
