# Project 2: Image Blurring 

```python
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Create or load a grayscale image
image = plt.imread('sample.jpg')  # Replace with your image path

# Convert RGB to grayscale if needed
if image.ndim == 3:
    R = image[:, :, 0]
    G = image[:, :, 1]
    B = image[:, :, 2]
    image = 0.299 * R + 0.587 * G + 0.114 * B

# Step 2: Get image dimensions
rows, cols = image.shape

# Step 3: Create an empty array for the blurred image
blurred = np.zeros((rows, cols))

# Step 4: Apply blur using a 3×3 average filter
for i in range(1, rows - 1):
    for j in range(1, cols - 1):
        # Take the 3×3 region around the current pixel
        region = image[i-1:i+2, j-1:j+2]
        # Replace pixel with the average of the region
        blurred[i, j] = np.mean(region)

# Step 5: Display the original and blurred images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.title("Original Grayscale Image")
plt.imshow(image, cmap='gray')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Blurred Image (3x3 Filter)")
plt.imshow(blurred, cmap='gray')
plt.axis('off')

plt.show()
```
