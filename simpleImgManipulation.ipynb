```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.image import imread

# 1. Load a real image instead of creating a fake one
fake_image = imread("sample.jpg")

# If the image is RGB (3 channels), convert it to grayscale
if fake_image.ndim == 3:
    # Weighted average to convert to grayscale
    fake_image = np.dot(fake_image[..., :3], [0.2989, 0.5870, 0.1140])

# 2. Transformations
# Invert the image colors
inverted_image = 1.0 - fake_image

# Crop the central part (from 25 to 75 pixels)
cropped_image = fake_image[25:75, 25:75]

# 3. Show the images
plt.figure(figsize=(10, 3))

plt.subplot(1, 3, 1)
plt.title("Original")
plt.imshow(fake_image, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("Inverted")
plt.imshow(inverted_image, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 3) # 1 row, 3 columns, third box
plt.title("Cropped")
plt.imshow(cropped_image, cmap='gray')
plt.axis('off')

plt.show()
```
